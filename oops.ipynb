# class is a logical entity/blue print or a prototype
# objects are instances of a class

```python
class Human:
    var=0
    def __init__(self,n,o):
        #two properties of the class
        self.name=n
        self.occupation=o

    #two methods of the class
    def do_work(self):
        if self.occupation=="tennis player":
            print(self.name,"plays tennis")
        elif self.occupation=="actor":
            print(self.name,"shoots film")

    def speaks(self):
        print(self.name,"syas how are you?")

#creating an instance of class        

tom=Human("tom cruise","actor")
tom.do_work()
tom.speaks()

maria=Human("maria sharapova","tennis player")
maria.do_work()
maria.speaks()


    
```

# Inheritance

Car and motorcycle are base class inherited from class vechile

# uses:

code reusability
extensibility
readbility


```python
class vehicle:
    def general_usage(self):
        print("general usage: transportation")

class car(vehicle):
    def __init__(self):
        print("this is a car")
        self.wheels=4
        self.has_roof=True

    def specific_usage(self):
        self.general_usage()
        print("specific use: commute to work, vacation with family")

class Motorcycle(vehicle):
    def __init__(self):
        print("this is a motorcycle")
        self.wheels=2
        self.has_roof=False

    def specific_usage(self):
        self.general_usage()
        print("road trip, racing")

c=car()        
c.general_usage()
c.specific_usage()    

#checking is an object is an instance of it or not
print(isinstance(c,car))

#checkinng if a class is an instance of another class (parent class)
print(issubclass(car,vehicle))
```

```python
#multiple inheritance
class Father():
    def skills(self):
        print("gardening,programming")

class Mother():
    def skills(self):
        print("cooking,art")

class child(Father,Mother):
    def skills(self):
        Father.skills(self)
        Mother.skills(self)
        print("sports")
c=child()
c.skills()                       
```

```python
#Encapsulation is wrapping data and functions as a single unit
# restricts direct access to some of an objects components

class BankAccount:
    def __init__(self,balance):
        self.__balance=balance   #private variable

    def deposit(self,amount):
        self.__balance+=amount

    def get_balance(self):
        return self.__balance

#Usage
account=BankAccount(1000)
account.deposit(500)
print(account.get_balance())
# print(account.__balance) 

#__balance is a private variable (cannot be accessed directly)
#deposit() and get_balance() providing us controlled access
```

```python
#Polymorphism allows different classes to use the same method name

class Bird:
    def fly(self):
        print("Bird can fly")
class Penguin(Bird):
    def fly(self):
        print("penguins cannot fly")

for obj in [Bird(),Penguin()]:
    obj.fly()
```

```python
#Abstraction hiding implementation details and exposing only necessary parts @abstract

from abc import ABC,abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        print("car starts with a key")


ab=Vehicle()
ab.start()
car=Car()
car.start()        
```
