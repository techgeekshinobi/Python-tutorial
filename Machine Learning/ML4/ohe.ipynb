```python
import pandas as pd
```

```python
df=pd.DataFrame({
    'Color':['Red','Green','Blue','Green']
})

df
```

```python
#One-hot-encoding
```

```python
encode_df=pd.get_dummies(df,columns=['Color'])
print(encode_df)
```

dummy variable trap:

one or more dummy variables are highly correlated, it introduces collinearity, which can distort regression models like linear regression
