We have a dataset containing prices of used bm2 cars. We are going to analyze this dataset and build a prediction function that can predict a price by taking mileage and age of the car as input. We will use sklean train_test_split method to split training and testing dataset.

```python
import pandas as pd
```

```python
df=pd.read_csv('carprices.csv')
df.head()
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
plt.scatter(df['Mileage'],df['Sell Price($)'])
plt.show()
```

```python
plt.scatter(df['Age(yrs)'],df['Sell Price($)'])
plt.show()
```

```python
X=df[['Mileage','Age(yrs)']]
X
```

```python
y=df['Sell Price($)']
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3)
```

```python
X_train
```

```python
X_train
```

```python
X_test
```

```python
y_train
```

```python
y_test
```

```python
from sklearn.linear_model import LinearRegression
clf=LinearRegression()
clf.fit(X_train,y_train)
```

```python
X_test
```

```python
clf.predict(X_test)
```

```python
y_test
```

```python
clf.score(X_test,y_test)
```
