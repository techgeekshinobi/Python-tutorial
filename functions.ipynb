```python
#Recursive functions,recursion 
#functions that calls iteslf within its body
#5
def factorial(n):
    if n==0:
        return 1
    return n*factorial(n-1)  # 5*fact(4) ---> 4*fact(3)--->3*fact(2)--->2*fact(1)--->1*fact(0)
factorial(5)                       # 120            24                6             2               1
```

```python
#Function can have parameters, return value, and default arguments
def add(a,b=15):   # b has a default value
    return a+b

print(add(5))  #using deafult value of b
# print(add(5,20))# ovverides default value b
```

```python
#variable length argument (*args)
def multiply(*args):
    result =1
    for num in args:
        result*=num
    return result    
multiply(2,3,4)    

#keyword length argument

def print_info(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
print_info(name="shubham",age=23)

```

```python
#lambda expressions
#Functions that do not have a name

square=lambda x: x*x
square(2)

square(5)
```

```python
# program to check whether a number is prime or not using function

def prime(n):
    is_prime=True
    for i in range(2,n):
        if n%i==0:
            is_prime=False
            break
    if is_prime:
        print("it is prime")
    else:
        print("it is not prime")        
prime(9)
```
