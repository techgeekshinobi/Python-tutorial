# sets
# used to store a collection of items
# no duplicate elements are allowed
# unordered collection , we cannot access items using indexes----> implemented uing hashing internally
# mutable but you cannot modify existing elements {1,2,3}

```python
s={1,2,3}
print(s)
type(s)
```

```python
#typecasting
s=set([1,2,3])
print(s)

s.add(4)
print(s)
```

```python
#immutable
s={"apple","banana","cherry"}
print(s)

# # s[1]="kiwi"
# print(s)

s.add("kiwi")
print(s)
```

```python

#frozenset
fs=frozenset([1,2,3])
```

```python
#union(|) and intersection(&)
#union() and itersection()

people1={"shubham","akash","lovepreet"}
people2={"shubham","harshit"}

#Union
print(people1|people2)
print(people1.union(people2))

#Inersection
print(people1&people2)
print(people1.intersection(people2))
```

```python
#differences of sets ('-')
set1={1,2,3,4,5}
set2={3,4,5,6,7}
print(set1-set2)
print(set1.difference(set2))
```

```python
#clear a set
set1={1,2,3,4,5}
set1.clear()
print(set1)
```
