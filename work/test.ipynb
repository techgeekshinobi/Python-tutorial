# identifying the duplicate and removing without parsing 

```python
num=[13,14,10,3,5,6,16,8,7,17]
for i in range(len(num)):
    for j in range(i+1,len(num)):
        if num[i]>num[j]:
            temp=num[i]
            num[i]=num[j]
            num[j]=temp
print(num)

#Bubble sort
#insertion sort 
```
