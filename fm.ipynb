# File modes
Modes--------Deescription

r---> Read(default),fails if the files doent exists
w----> Write(creates a new file or truncates an existing one)
a-----> Append(create a file if doesn't exists, writes at the end)


r+---> Read and write
w+----> write and read
a+---> append and read

```python
# Reading a file 
with open("file.txt","r") as file:
    content=file.read()
    print(content)
```

```python
# Writing to a file
with open("file.txt","w") as file:
    file.write("hello there")
```

```python
#Appending to a file
with open("file.txt",'a') as  file:
    file.write("\n this is a new line i added")
```

```python
#Reading and writing
with open("file.txt","r+") as file:
    print(file.read())
    file.write("\this is the third new line i added")
```
