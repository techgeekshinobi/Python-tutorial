```python
import numpy as np
```

```python
a=np.arange(12).reshape(3,4)
a
```

```python
for row in a:
    for cell in row:
        print(cell)
```

```python
for cell in a.flatten():
    print(cell)
```

```python
#numpy provides nditer which is kind of a sophisticated way of doing iteration
```

```python
for x in np.nditer(a,order='C'):
    print(x)
```

```python
for x in np.nditer(a,order='F'):
    print(x)
```

```python
for x in np.nditer(a,order='F',flags=['external_loop']):
    print(x)
```

```python
for x in np.nditer(a,op_flags=['readwrite']):
    x[...]=x*x
```
