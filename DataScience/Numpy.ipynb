# three main benefits that numpy array has over python list

1. Less memory
2. Fast
3. Convenient

```python
#Less memory

import sys
import numpy as np

l=range(1000)
print(sys.getsizeof(10)*len(l))


array=np.arange(1000)
print(array.size*array.itemsize)
```

```python
#Fast

import time
import numpy as np

SIZE=1000000

l1=range(SIZE)#[0,1,2,3...]    [(0,0),(1,1),(2,2),(3,3)....]
l2=range(SIZE)#[0,1,2,3....]

a1=np.arange(SIZE)
a2=np.arange(SIZE)

start=time.time()
result=[(x+y) for x,y in zip(l1,l2)]
print("python list took:",(time.time()-start)*1000)


start=time.time()
result=a1+a2
print("numpy took:",(time.time()-start)*1000)
```
