```python
#creating a list
fruits=["1","banana","cherry","date","elderberry"]
print(fruits)
```

```python
#accessing elements in a list
fruits=["apple","banana","cherry","date","elderberry"]
print(fruits[0])
print(fruits[1])
print(fruits[-1])
```

```python
#Modifying a list
fruits=["apple","banana","cherry","date","elderberry"]
fruits[1]="blackberry"
print(fruits)
```

In java or c++ arrays are static that means their sizes are fixed one defined

```python
#append function
fruits=["apple","banana"]
fruits.append("cherry")
print(fruits)
```
