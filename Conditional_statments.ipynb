# If statement

```python
a=1
if(a<2):
    print("smaller number")
```

```python
#if -else conditional
age=18
if(age>18):
    print("eligible for vote")
else:
    print("not eligible to vote")    
```

```python
#short hand if-else
#Also known as Ternary conditional assignment
marks=45
res="pass" if marks>40 else "fail"
print(res)
```

```python
#elif
age=31
if age<=12:
    print("child")
elif age<=19:
    print("teen")
elif age<=30:
    print("young adult")
else:
    print("adult")
```

```python
#Nested if
age=70
member=True

if age>=60:
    if member:
        print("discount available")
    else:
        print("no discount")
else:
    print("no discount for this age group")        
```

```python
#match case --> switch case
number=2
match number:
    case 1:
        print("one")
    case 2 | 3:
        print("two or three")
    case _:
        print("default case")

  
```
