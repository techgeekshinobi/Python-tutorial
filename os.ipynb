# file operations using os module

```python
import os

#rename
os.rename("newfile.txt","newestfile.txt")
```

```python
#delete file
os.remove("newestfile.txt")
```

# directory operations


```python
os.mkdir("new_directory")
```

```python
#creating muktiple nested directories
os.makedirs("parent_dir/child_dir")
```

```python
#remove an empty directory
os.rmdir("new_directory")
```

```python
os.removedirs("parent_dir/child_dir")
```

```python
#listing files and directories in a given path
os.listdir(".")
```

```python
#path operations 

#to check if path exists or not
os.path.exists("lis.ipynb")
```

```python
# check if the path is a file or a directory
os.path.isfile("list.ipynb")
os.path.isdir("new_directory")
```

```python
#get a file name or directory nsme from a path
os.path.basename("loop.ipynb")
```
