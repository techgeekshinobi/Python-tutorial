# Dictionaries
# keys can't be repeated and must be immutable
# internally using hashing so takes contant amount of time 

```python
#create a dictionary
dict1={1:"appple",2:"banana",3:"orange"}
print(dict1)

dict2=dict(a="appple",b="banana",c="orange")
print(dict2)
```

```python
#Accessing items in a dictionary
d={"name":'shubham',1:"python",(1,2):[1,2,3]}
print(d[(1,2)])
```

```python
#add 
d={1:"red",2:"blue",3:"green"}
d[4]="yellow"

#update
d[1]="gray"
d
```

```python
#remove items from dictionary
d={1:"red",2:"blue",3:"green"}

#using del to remove an element
del d[1]
d
```

```python
#using pop() remove an item by key and return its value
d={"name":"red",2:"blue",3:"green"}
d.pop("name")
```

```python
#using popitem() ---> removes and returns the last key-pair
d={1:"red",2:"blue",3:"green"}
d.popitem()
```

```python
#iterating through a dictionary
d={1:"red",2:"blue",3:"green"}

#over key iteration
for key in d:
    print(key)

#ove value iteration
for values in d.values():
    print(values)

for i,j in d.items():
    print(f"{i}:{j}")
```

```python
#nested dictionaries
d={1:"hello",2:"good",3:{1:"red",2:"blue",3:"green"}}

for key,value in d.items():
    if key==3:
        demo_dict=value
        for i,j in demo_dict.items():
            print(key,".",i,j)
    else:
        print(key,value)        



    
        
        
```

```python
#adding using methods
d={1:10,2:20,3:30}
d.update({4:40,5:50})
print(d)
```

```python
#adding items from an iterable
d={1:10,2:20,3:30}
a=[[4,40],[5,50]]
d.update(a)
print(d)
```

```python
#using setdefault() method
#if key exists it returns value else it adds the key with the specific value
d={1:10,2:20,3:30}
d.setdefault(4,40)
```

```python
#clearing a dictionary
d={1:10,2:20,3:30}
d.clear()
print(d)
```

```python
d={1:10,2:20,3:30}
d1=d.copy()
print(d1)
```

```python
#createa new dictionary with keys from the iterable and values set to value
keys={"a","b","c","d"}
d=dict.fromkeys(keys,0)
print(d)
```
