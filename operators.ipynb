operators and operand

```python
a=15
b=4

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
print(a**b)

```

```python
#Relational operators always yields a boolean value
a=13
b=33

print(a>b)
print(a<b)
print(a==b)
print(a!=b)
print(a>=b)
print(a<=b)
```

```python
#Logical operators results in boolean value
a=True
b=False
print(a and b)
print(a or b)
print(not a)
```
